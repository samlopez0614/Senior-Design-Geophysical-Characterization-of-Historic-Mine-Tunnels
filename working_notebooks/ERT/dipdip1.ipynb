```python
file = "../../Data/ERT data/DipoleDipole_1-2023-03-07-214954.dat"

with open(file, "r") as f:
    lines = f.readlines()

for i, line in enumerate(lines[:15]):
    print(i, line.strip())
```

```python
n_data = int(lines[6].strip())

print("Number of measurements:", n_data)
```

```python
data = []

for line in lines[9:9+n_data]:

    values = list(map(float, line.split()))

    array_code = values[0]

    Ax = values[1]
    Az = values[2]

    Bx = values[3]
    Bz = values[4]

    Mx = values[5]
    Mz = values[6]

    Nx = values[7]
    Nz = values[8]

    R = values[9]

    data.append([
        Ax, Az,
        Bx, Bz,
        Mx, Mz,
        Nx, Nz,
        R
    ])
```
