## Dipole-Dipole Survey
### Howelson Hill, Steamboat Springs, CO

This notebook is processing and making prelimenary reports from the data file 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = "../../Data/ERT data/DipoleDipole_2-2023-03-07-214927.dat"

# There are 9 header lines before the measurements
data = np.loadtxt(filename, skiprows=9, max_rows=1980)

print(data.shape)
print(data[:5])
```

```python
# create data frame
df = pd.DataFrame(
    data,
    columns=[
        "array",
        "Ax", "Az",
        "Bx", "Bz",
        "Mx", "Mz",
        "Nx", "Nz",
        "R"
    ]
)

df.head()
```
